## Step 9: Make new columns

`np.where` works like an IF formula in Excel: High if above the median, otherwise Low.

```python
df['Amount_Level'] = np.where(amount > amount.median(), 'High', 'Low')
```

```python
df['Amount_Level'].value_counts()
```

Get the year and month from the date

```python
df['Year'] = df['Transaction_Date'].dt.year
```

```python
df['Month'] = df['Transaction_Date'].dt.month
```

```python
df[['Transaction_Date', 'Year', 'Month', 'Amount_Level']].head()
```

## Step 10: Charts

### Chart 1: Number of transactions by type (bar chart)

```python
plt.figure(figsize=(9, 5))
sns.countplot(data=df, x='Transaction_Type')
plt.title('Number of Transactions by Type')
plt.xlabel('Transaction Type')
plt.ylabel('Count')
plt.xticks(rotation=30)
plt.show()
```

### Chart 2: Average amount by channel (bar chart)

```python
plt.figure(figsize=(9, 5))
sns.barplot(data=df, x='Channel', y='Transaction_Amount_INR', errorbar=None)
plt.title('Average Transaction Amount by Channel')
plt.xticks(rotation=30)
plt.show()
```

### Chart 3: Transaction status share (pie chart)

```python
status_counts = df['Transaction_Status'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(status_counts, labels=status_counts.index, autopct='%1.1f%%')
plt.title('Transaction Status Share')
plt.show()
```

### Chart 4: Distribution of amounts (histogram)
We cut off the top 1% of values so a few huge amounts do not squash the chart.

```python
limit = np.percentile(amount, 99)
plt.figure(figsize=(9, 5))
sns.histplot(amount[amount <= limit], bins=50)
plt.title('Distribution of Transaction Amount')
plt.xlabel('Amount (INR)')
plt.show()
```

### Chart 5: Total amount per month (line chart)

```python
monthly = df.groupby(['Year', 'Month'])['Transaction_Amount_INR'].sum().reset_index()
monthly['Date'] = pd.to_datetime(monthly[['Year', 'Month']].assign(day=1))
monthly.head()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(monthly['Date'], monthly['Transaction_Amount_INR'])
plt.title('Total Transaction Amount per Month')
plt.xlabel('Month')
plt.ylabel('Total Amount (INR)')
plt.show()
```

### Chart 6: Amount by city (box plot)

```python
plt.figure(figsize=(11, 5))
sns.boxplot(data=df, x='Customer_City', y='Transaction_Amount_INR', showfliers=False)
plt.title('Amount by City (extreme values hidden)')
plt.xticks(rotation=30)
plt.show()
```

### Chart 7: Status by channel (stacked bar)

```python
status_by_channel = pd.crosstab(df['Channel'], df['Transaction_Status'], normalize='index')
status_by_channel.plot(kind='bar', stacked=True, figsize=(9, 5))
plt.title('Status Share by Channel')
plt.ylabel('Share')
plt.xticks(rotation=30)
plt.show()
```

### Chart 8: Correlation heatmap
Correlation goes from -1 to +1. Values near 0 mean the columns are not related.

```python
number_columns = ['Transaction_Amount_INR', 'Balance_After_INR', 'Interest_Rate_Pct', 'Loan_Amount_INR']
plt.figure(figsize=(7, 5))
sns.heatmap(df[number_columns].corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Between Number Columns')
plt.show()
```
